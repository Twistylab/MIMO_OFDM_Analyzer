Рассмотрим способ эквализации сигнала с переменными весами обоих фильтров (_"Digital Communications and Signal Processing" by K Vasudevan, section 5.1.8_). Для этого распишем эквализированный сигнал через вектора весов, сигнала на выходе из канала и ранее детектированных символов:
$$
y_k = {\vec{w}^{ff}}^T \cdot \vec{\nu} - {\vec{w}^{bf}}^T \cdot \vec{\hat{I}} = f_k - b_k,
$$
где
$$
\begin{aligned}
& \vec{\nu} = \{\nu_i\}_{i=k}^{k-K_1+1}\\
& \vec{\hat{I}} = \{\hat{I}_i\}_{i=k-1}^{k-K_2+1}\\
& \vec{w}^{ff} = \{w^{ff}_i\}_{i=0}^{K_1-1}\\
& \vec{w}^{bf} = \{w^{bf}_i\}_{i=0}^{K_2-1}
\end{aligned}
$$
Введём функцию стоимости ошибки MSE (**Mean Squared Error**) и саму ошибку:
$$
J = \mathbb{E}[e_k^2]; \quad e_k = y_k - I_k .
$$
Через градиентный спуск будем минимизировать $J$ и обновлять веса фильтров:
$$
\vec{w} := \vec{w} - \mu\nabla J,
$$
где $\mu$ - шаг обучения.

Найдём $\nabla J$:
$$
\nabla J = \nabla \mathbb{E}[e_k^2] = 2\mathbb{E}[e_k \nabla e_k].
$$
Распишем градиент ошибки через веса фильтров:
$$
\nabla e_k = \nabla (y_k - I_k) = \left\{ \begin{aligned}
& \frac{\partial y_k}{\partial \vec{w}^{ff}} = \vec{\nu} \\
& \frac{\partial y_k}{\partial \vec{w}^{bf}} = -\vec{\hat{I}}
 \end{aligned}\right.
$$
Подставим данные выражения в градиентный спуск, перейдя от математического ожидания к мгновенным значениям:
$$
\begin{aligned}
& \vec{w}^{ff} := \vec{w}^{ff} - \mu e_k\vec{\nu} \\
& \vec{w}^{bf} := \vec{w}^{bf} + \mu e_k \vec{\hat{I}}
 \end{aligned},
$$
где $\mu$ - шаг обучения, определяется он следующим образом:
$$
0 < \mu < \mu_{\max} = \frac{2}{\lambda_{\max}} = \frac{2}{K_1\mathbb{E}[|\vec{\nu}|^2] + K_2\mathbb{E}[|\hat{I}_k|^2]},
$$
где
$$
\mathbb{E}[|\vec{\nu}|^2] = \frac{1}{K_1}\sum_{i=0}^{K_1-1} |\vec{\nu}_i|^2; \quad
\mathbb{E}[|\vec{\hat{I}}|^2] = \frac{1}{K_2}\sum_{i=0}^{K_2-1} |\vec{\hat{I}}_i|^2;
$$

Определим шаг обучения. Пусть $\vec{w}^{ff} \equiv \vec{w}$, тогда:
$$
\begin{gathered}
J = \mathbb{E}[(f_k - I_k)^2] = \mathbb{E}[(\vec{w}^T\vec{\nu} - I_k)^2] = \mathbb{E}[(\vec{w}^T\vec{\nu})^2] - 2\mathbb{E}[I_k\vec{w}^T\vec{\nu}] + \mathbb{E}[I_k^2] = \\ \\
= \vec{w}^T\mathbb{E}[\vec{\nu}\vec{\nu}^T]\vec{w} - 2\vec{w}^T \mathbb{E}[I_k\vec{\nu}] + \mathbb{E}[I_k^2] = \vec{w}^T R \vec{w} - 2\vec{w}^T\vec{p} + \mathbb{E}[I_k^2].
\end{gathered}
$$
Запишем условие минимума:
$$
\vec{w}^* = \text{argmin}(J) \quad \Rightarrow \quad \nabla J = 0,
$$
тогда:
$$
\vec{w}^*=R^{-1}\vec{p}.
$$
Введём ошибку весов:
$$
\vec{\tilde{w}} = \vec{w} - \vec{w}^* .
$$
Подставим в градиентный спуск:
$$
\vec{\tilde{w}} := \vec{\tilde{w}} - \mu e_k \vec{\nu}.
$$
Выразим ошибку через $\vec{\tilde{w}}$:
$$
e_k = \vec{w}^T\vec{\nu} - I_k \quad \Rightarrow \quad e_k = \vec{\tilde{w}}^T\vec{\nu} + \vec{w}^*\vec{\nu} - I_k.
$$
Введём оптимальную ошибку:
$$
e_k^* = \vec{w}^*\vec{\nu} - I_k \quad \Rightarrow \quad e_k = e_k^* + \vec{\tilde{w}}^T\vec{\nu}.
$$
Подставляем:
$$
\vec{\tilde{w}} := \vec{\tilde{w}} - \mu e_k \vec{\nu} = \vec{\tilde{w}} - \mu (e_k^* + \vec{\tilde{w}}^T\vec{\nu}) \vec{\nu}.
$$
Возьмём мат. ожидание:
$$
\mathbb{E}[\vec{\tilde{w}}] := \mathbb{E}[\vec{\tilde{w}}] - \mu\mathbb{E}[e_k^*\vec{\nu}] - \mu\mathbb{E}[\vec{\nu}\vec{\nu}^T\vec{\tilde{w}}].
$$
При условии оптимальности ошибки:
$$
\mathbb{E}[e_k^*\vec{\nu}] = 0
$$
из-за ортогональности. Тогда:
$$
\mathbb{E}[\vec{\tilde{w}}] := (I - \mu R)\mathbb{E}[\vec{\tilde{w}}].
$$
